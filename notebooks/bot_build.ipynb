# Dataset Loading


```python
Path.as_posix()
```

```python
from langchain_community.document_loaders import PyMuPDFLoader

loader = PyMuPDFLoader(
    file_path=Path.as_posix(),
    mode="page",
    extract_images=True,
    images_inner_format="html-img",
    extract_tables="markdown"
)

pages = loader.load()

print(f"Total pages loaded: {len(pages)}")
```

```python
pp(pages[154].page_content) 
```

```python
pp(pages[125].metadata)
```

# Chunking

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter  
```

```python
splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50,
)
```

```python
chunking = splitter.split_documents(pages)
```

```python
print(f"Total chunks created: {len(chunking)}")
```

```python
print(pages[27].page_content)
```

```python
print(chunking[153].page_content)
```

```python
print(chunking[155].page_content)
```

# Embeddings

```python
from langchain_huggingface import HuggingFaceEmbeddings    
```

```python
embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2",
    encode_kwargs={"normalize_embeddings": True},
)
document_embeddings = embeddings.embed_documents(
    texts=[chunk.page_content for chunk in chunking]
)
```

```python
len(document_embeddings[0])
```

# Storing Embedding in Vector Stores


```python
Path
```

```python
persist_directory = Path / "notebooks" / "embeddings"
```

```python
collection_name = "rag-pipeline"
```

```python
from langchain_chroma import Chroma
```

```python
collection_name
```

```python
persist_directory
```

```python
vectorstore = Chroma.from_documents(
    documents=chunking,
    embedding=embeddings,
    persist_directory=persist_directory,
    collection_name=collection_name
)

print(f"Vectorstore created with {vectorstore._collection.count()} vectors.")
```

```python
print(f"Stored {len(document_embeddings)} chunks in the '{collection_name}' collection.")
```
